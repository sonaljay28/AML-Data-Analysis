```python
import pandas as pd
```

```python
import numpy as np
```

```python
class BankingSystem:
    def __init__(self):
        self.accounts = pd.DataFrame(columns=['Account Number', 'Balance'])

    def create_account(self, account_number, initial_balance):
        self.accounts.loc[len(self.accounts)] = [account_number, initial_balance]

    def deposit(self, account_number, amount):
        account = self.accounts[self.accounts['Account Number'] == account_number]
        if len(account) > 0:
            account.loc[0, 'Balance'] += amount
        else:
            print("Account not found")
```

```python
def withdraw(self, account_number, amount):
        account = self.accounts[self.accounts['Account Number'] == account_number]
        if len(account) > 0:
            if account.loc[0, 'Balance'] >= amount:
                account.loc[0, 'Balance'] -= amount
            else:
                print("Insufficient funds")
        else:
            print("Account not found")
```

```python
def get_balance(self, account_number):
        account = self.accounts[self.accounts['Account Number'] == account_number]
        if len(account) > 0:
            return account.loc[0, 'Balance']
        else:
            return None
```

```python
 def get_transaction_history(self, account_number):
        account = self.accounts[self.accounts['Account Number'] == account_number]
        if len(account) > 0:
            return account.loc[0, 'Balance']
        else:
            return None
```
